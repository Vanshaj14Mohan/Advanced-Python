# The random module

```python
# This module implements pseudo-random number generators for various distributions.

import random

# random float in [0,1)
a = random.random() # Print random float between 0 and 1
print(a)

# random float in range [a,b]
a = random.uniform(1,10) # adding specific range to it
print(a)

# To use integers
b = random.randint(1, 10) # giving a range
print(b)

c = random.randrange(1, 10) # upper bound won't be included
print(c)

# For statistics usage
d = random.normalvariate(0,1) # mean of 0 and stand dev of 1
print(d)
```

```python
# Different functions we could work with in random module
nums = list("ABCDEFGH")
print(nums)

# And if we need to pick a random choice from nums:
choice = random.choice(nums)
print(choice)

# And if we want sample data in it
sample = random.sample(nums, 4)
print(sample)
```
