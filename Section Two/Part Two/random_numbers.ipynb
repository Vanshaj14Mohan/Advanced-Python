# The random module

```python
# This module implements pseudo-random number generators for various distributions.

import random

# random float in [0,1)
a = random.random() # Print random float between 0 and 1
print(a)

# random float in range [a,b]
a = random.uniform(1,10) # adding specific range to it
print(a)

# To use integers
b = random.randint(1, 10) # giving a range
print(b)
```
